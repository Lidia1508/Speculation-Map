```python
import numpy as np
import matplotlib.pyplot as plt
import geopandas as gpd
import pandas as pd
import csv
```

```python
df=gpd.read_file("divisions-administratives-v2r1-municipis-1000000-20250730.json")
```

```python
df
```

```python
df.plot()
plt.show()
```

```python
dh=gpd.read_file("Impost_sobre_els_habitatges_buits_(IHB)_per_municipi_i_exercici_de_meritació_20261118.csv")
```

```python
dh['CODIMUNI'] = dh['Codi']
dh
```

```python
df = df.merge(dh, on='CODIMUNI', how='left' )
```

```python
df
```

```python
totqls = df.groupby("CODIMUNI").agg(
    Hab=("Habitatges","sum")).reset_index()
```

```python
totqls
```

```python
df_final = df.merge(totqls, on='CODIMUNI', how = 'left', suffixes=("", "_total"))
```

```python
df_final
```

```python
df_final.plot(column='Habitatges',legend=True)
plt.show()
```

```python
fig,axs=plt.subplots(1,1,figsize=(11,5))
df_final.plot(column='Habitatges',ax=axs)
plt.show()
```
